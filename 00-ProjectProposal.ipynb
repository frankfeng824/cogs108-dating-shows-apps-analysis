## Research Question

How is U.S. reality dating show popularity, measured by average per-episode ratings and per-season viewership, associated with changes in U.S. dating app activity (Tinder, Bumble, and Hinge), measured by annual revenue and annual user counts, from 2019–2026 using all available data?

## Background and Prior Work

The things that happen in dating shows are not like actual dating. As a form of reality TV that wishes to attrach a large, engaged audience, it's inevitable that producers and writers of dating shows artificialy infuse as much drama and tension as possible into each episode and season. Therefore, it shouldn't be a surprise that frequent viewers of dating shows tend to develop quite a warped perception of what romance should look like in everyday life. As early as 2007, a study already found that greater viewing of dating shows strongly predicts the tendencies to believe that "men are sex-driven," "dating is a game," and "women are sex objects."<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Unfortunately, such effects of dating shows have not changed much. A study looking at a much more recent dating show, "Singles Inferno," also found that the show perpetuates gender sterestypes and unrealistic beauty standards.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

Of course, dating apps, such as Tinder, Hinge, and Bumble, have also played quite a large role in the modern dating scene. The Pew Research Center finds that 30% of Americans have used dating apps at least once, and this number is particular higher for younger and queer people.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) However, just like dating shows, dating apps have also proven to be quite problematic. The same Pew Research Center report finds that over half of young women (18-34) report negative interactions while using dating apps, such as being sent unsolicited sexual content. Other studies find that dating app usage correlates with substance abuse and insecurity about physical appearances.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

Our project is motivated by the connections we see between dating shows and dating apps. They seem to both be related to the sexualization of women, superficial perceptions of romantic connection, and the perpetuation of unrealistic beauty standards. Although we are unable to run an experiment and establish any causal mechanisms between the two, we sense that the similar sociocultural sentiments underlying both dating shows and dating apps should be reflected by a correlation between dating show viewership/ratings and dating app usage. Beyond the aforementioned studies that look at either dating shows or dating apps seperately, we find very little prior empirical research on this topic. However, others have also previously speculated that both dating shows and dating apps (particularly Love Island and Tinder, respectively) collectively contribute to how intimacy has become more "performative" and "transactional" in the modern day.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) 

Finally, it's useful to describe the specific dating shows we decided to include. Although their inclusion was mostly guided by the fact that they had the most accessible viewership and ratings data, our final set of three turned out to be quite diverse. The first is Love Island USA, where multiple singles live together in a villa, couple up, get "dumped" by public vote, and reshuffle constantly as new members get introduced. The second, The Bachelor, is much more explicitly competitive, since its premise is that one person dates dozens of other contestants simulatenously. The third, Love is Blind, is also special, since participants aren't able to see one another and must rely on personality alone to assess each other, hence removing physical appearance from the equation.

## Hypothesis


We expect a positive relationship between the popularity of U.S. reality dating shows and dating app activity in the U.S. from 2019 to 2026. Specifically, when a season has higher average viewership and ratings, we predict that user counts and annual revenue for apps like Tinder, Bumble, and Hinge will also increase in the same year or shortly after.

We make this prediction because dating shows and dating apps seem to have similar underlying sociocultural sentiments, as mentioned in the Background section. Additionally, dating shows may make dating more visible and normalized to viewers, making them wish to personally engage in dating behavior themselves. 

However, we do not expect this relationship to be perfectly and consistently apparent across all years, since there are other existing factors like economic conditions or app-specific changes, which may also affect user growth and revenue.

It should also include each dating app's (Tinder, Hinge, Bumble) monthy revenue and user counts/increase.
