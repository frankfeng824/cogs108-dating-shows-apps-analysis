## Research Question

How is U.S. reality dating show popularity, measured by average per-episode ratings and per-season viewership, associated with changes in U.S. dating app activity (Tinder, Bumble, and Hinge), measured by annual revenue and annual user counts, from 2019–2026 using all available data?



## Background and Prior Work

The things that happen in dating shows are not like actual dating. As a form of reality TV that wishes to attrach a large, engaged audience, it's inevitable that producers and writers of dating shows artificialy infuse as much drama and tension as possible into each episode and season. Therefore, it shouldn't be a surprise that frequent viewers of dating shows tend to develop quite a warped perception of what romance should look like in everyday life. As early as 2007, a study already found that greater viewing of dating shows strongly predicts the tendencies to believe that "men are sex-driven," "dating is a game," and "women are sex objects."<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Unfortunately, such effects of dating shows have not changed much. A study looking at a much more recent dating show, "Singles Inferno," also found that the show perpetuates gender sterestypes and unrealistic beauty standards.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

Of course, dating apps, such as Tinder, Hinge, and Bumble, have also played quite a large role in the modern dating scene. The Pew Research Center finds that 30% of Americans have used dating apps at least once, and this number is particular higher for younger and queer people.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) However, just like dating shows, dating apps have also proven to be quite problematic. The same Pew Research Center report finds that over half of young women (18-34) report negative interactions while using dating apps, such as being sent unsolicited sexual content. Other studies find that dating app usage correlates with substance abuse and insecurity about physical appearances.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

Our project is motivated by the connections we see between dating shows and dating apps. They seem to both be related to the sexualization of women, superficial perceptions of romantic connection, and the perpetuation of unrealistic beauty standards. Although we are unable to run an experiment and establish any causal mechanisms between the two, we sense that the similar sociocultural sentiments underlying both dating shows and dating apps should be reflected by a correlation between dating show viewership/ratings and dating app usage. Beyond the aforementioned studies that look at either dating shows or dating apps seperately, we find very little prior empirical research on this topic. However, others have also previously speculated that both dating shows and dating apps (particularly Love Island and Tinder, respectively) collectively contribute to how intimacy has become more "performative" and "transactional" in the modern day.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) 

Finally, it's useful to describe the specific dating shows we decided to include, which were Love Island USA, The Bachelor, and Love is Blind. There were primarily two considerations that drove our selection of them. The first is practical, since they were the shows that turned out to have the most available data. Other shows rarely had sufficient data on viewership, which would have made them extremely difficult to include in our analysis. Secondly, we believe that the diversity of these three shows will eventually help us interpret our results more effectively. Our hypothesis that dating show viewership or ratings (or both) can predict dating app usage is founded on the observation that they reflect similar sociocultural sentiments. However, not all dating shows reflect identical sentiments and values. Specifically, Love Island USA, where multiple singles live together in a villa, couple up, get "dumped" by public vote, and reshuffle constantly, appears to magnify how dating is messy and full of drama. In contrast, The Bachelor magnifies how dating is competetive, since its premise is that one person dates and choose among dozens of other contestants simulatenously. Finally, the third show, Love is Blind, is also special, since participants aren't able to see one another and must rely on personality alone to assess each other, removing physical appearance and superficiality from the equation. Therefore, if we were to find correlations between certain shows and dating app usage, but not the same correlations for other dating shows, then there is an argument to be made that the values underlying those shows, specifically, are more tied to dating app usage.

## Hypothesis


We expect a positive relationship between the popularity of U.S. reality dating shows and dating app activity in the U.S. from 2019 to 2026. Specifically, when a season has higher average viewership and ratings, we predict that user counts and annual revenue for apps like Tinder, Bumble, and Hinge will also increase in the same year. 

We make this prediction because dating shows and dating apps seem to have similar underlying sociocultural sentiments, as mentioned in the Background section. Additionally, dating shows may make dating more visible and normalized to viewers, making them wish to personally engage in dating behavior themselves. 

However, we do not expect this relationship to be perfectly and consistently apparent across all years, since there are other existing factors like economic conditions or app-specific changes, which may also affect user growth and revenue.

### Data Set 2: Viewership of Shows
Description: This dataset contains the number of viewers per season for the dating shows, Love Island USA, Love is Blind and The Bachelor  in millions from 2019 to 2025. The rough benchmark for average number of viewers per season for a dating show is 2.5 million. Values >2.5 million viewers per season indicate a larger audience and mainstream popularity. Values <2.5 million viewers per season indicate a smaller audience base and less commercially success. Values <1 million viewers per season are an indicator of a commercial flop and failure to reach audiences. The data for Love Island USA and Love is Blind were obtained from Netflix's Engagament Report (https://assets.ctfassets.net/4cd45et68cgf/1HyknFM84ISQpeua6TjM7A/97a0a393098937a8f29c9d29c48dbfa8/What_We_Watched_A_Netflix_Engagement_Report_2023Jan-Jun.xlsx) and the data for The Bachelor was obtained from TVSeriesFinale Website (https://www.google.com/url?sa=t&source=web&rct=j&opi=89978449&url=https://tvseriesfinale.com/tv-show/the-bachelor-season-29-ratings/&ved=2ahUKEwjl3cTfjLWUAxXrIEQIHbpqPAQQFnoECBoQAQ&usg=AOvVaw0OR86j6AWmgIxXY97c7S3h)

Major Concerns: The Bachelor has the number of viewers per episode released for each season and hence, the number of viewers per season is calculated as the average of the number of viewers per episode. However, Netflix’s annual engagement report from 2023 contains the number of hours each season of Love Island USA and Love is Blind have been viewed. In order to standardize the variable used for comparison, the standardized estimate of viewers per season for Love Island USA and Love is Blind have been calculated using the following formula:

**Number of viewers = Number of Hours Viewed / Total Runtime of the Season**

We acknowledge that this is an imperfect measurement, because we should not assume that all viewers watched all episodes of the entire season. However, Netflix has strictly released only in the hours watched for each season of a given show and no other data. Since obtaining perfect data is unrealistic due to Netflix's policy, we have decided to continue our project with this approxiate standardization  

Only for this dataset, we decided use the Netflix Report and TVSeriesFinaleReport directly in the raw form to demonstrate what type of data is available to us. The Netflix Runtime data, shows rating data, dating apps' revenue and user data were compiled into CSVs by transforming the raw data into a readable format.

```python
bachelor_rawdata = pd.read_csv('data/00-raw/BachelorViewership.csv') # loading the raw data for The Bachelor
print(bachelor_rawdata.head())
print("\nThe shape of the raw data for The Bachelor is", bachelor_rawdata.shape)
```

```python
bachelor_views =  bachelor_rawdata[bachelor_rawdata['Air Day']=='Season averages']
bachelor_views.reset_index(inplace=True)
print(bachelor_views)
print('Number of missing values per column')
print(bachelor_views.isna().sum())
```

Currently, the data contains some missing values in the Air Date and Episode columns, but we do not need this data for our analysis. We only need the season number and the associated season average number of viewers (in millions) for each season.

```python
bachelor_views = bachelor_views[['Season','Viewers (millions)','Year']]
bachelor_views  = bachelor_views.rename(columns={'Season':'season_num','Viewers (millions)': 'num_viewers (mil)','Year':'year'})
bachelor_views['show_name']='The Bachelor'
bachelor_views['above_2.5mil']=(bachelor_views['num_viewers (mil)']>2.5)
print(bachelor_views)
print("\nThe shape of the viewership data for The Bachelor is", bachelor_views.shape)
```

```python
bachelor_views.isna().sum() # checking the total number of missing values
```

The dataset is already tidy based on observation. Each row represents a specific season, the year, the average number of viewers (in millions) and whether the average number of viewers is above 2.5 million or not as a boolean value for that season of The Bachelor. There are no missing values remaining in the data.

Summary statistics for The Bachelor's viewership data are shown below.

```python
bachelor_views['num_viewers (mil)'].describe() # general summary statistics
```

```python
netflix_report = pd.read_csv('data/00-raw/What_We_Watched_A_Netflix_Engagement_Report_2023Jan-Jun.csv') # loading the raw data for The Bachelor
print(netflix_report.head())
print("\nThe shape of the Netflix Report is", netflix_report.shape)
```

```python
print('Number of missing values per column')
print(netflix_report.isna().sum())
```

There are some missing values in the Release Date column, but this column is not needed for analysis since the year of airing is to be obtained from the Netflix runtime data.

```python
target_rows = ['Love Is Blind: Season 1','Love Is Blind: Season 2','Love Is Blind: Season 3','Love Is Blind: Season 4','Love Island USA: Season 1','Love Island USA: Season 2','Love Island USA: Season 3','Love Island USA: Season 4']
netflix_report = netflix_report[netflix_report['Title'].isin(target_rows)]
netflix_report = netflix_report.drop(columns={'Available Globally?','Release Date'})
netflix_report.reset_index(inplace=True)
print(netflix_report)
print("\nThe shape of the filtered Netflix Report is", netflix_report.shape)
```

```python
netflix_runtime = pd.read_csv('data/00-raw/NetflixRuntime.csv') # loading the raw data for netflix runtime data
print(netflix_runtime)
print("\nThe shape of the Netflix Runtime data is", netflix_runtime.shape)
```

```python
print('Number of missing values per column')
print(netflix_runtime.isna().sum())
```

The dataset contains no missing values and hence, can be filtered further without any cleaning.

```python
netflix_runtime['Runtime (hours)'] = (netflix_runtime['Runtime (minutes)']/60).round(2)
netflix_runtime = netflix_runtime.drop(columns=['Runtime (minutes)'])
print(netflix_runtime)
```

```python
netflix_merged = netflix_report.merge(netflix_runtime, on='Title')
print(netflix_merged)
```

```python
netflix_merged['Hours Viewed'] = netflix_merged['Hours Viewed'].str.replace(',','').astype(float)
netflix_merged['num_viewers (mil)']=(netflix_merged['Hours Viewed']/netflix_merged['Runtime (hours)'])/1000000
netflix_merged['num_viewers (mil)'] = (netflix_merged['num_viewers (mil)']).round(3)
netflix_merged = netflix_merged.drop(columns={'Hours Viewed','Runtime (hours)'})
print(netflix_merged)
```

Summary statistics for Love is Blind and Love Island USA viewership data are shown below.

```python
netflix_merged.isna().sum() # checking the total number of missing values
```

The dataset is already tidy based on observation. Each row represents a specific season for specifc show, the year and the average number of viewers (in millions) for that season. There are no missing values remaining in the data.

```python
netflix_merged[['show_name','season_num']]= netflix_merged['Title'].str.split(': Season',expand=True)
netflix_merged = netflix_merged.drop(columns=['Title'])
netflix_merged = netflix_merged.rename(columns={'Year':'year'})
print(netflix_merged)
```

```python
netflix_merged['above_2.5mil']=(netflix_merged['num_viewers (mil)']>2.5)
print(netflix_merged)
print("\nThe shape of the filtered Netflix data is", netflix_merged.shape)
```

```python
netflix_merged['num_viewers (mil)'].describe() # general summary statistics
```

```python
show_viewership = pd.concat([bachelor_views, netflix_merged], axis=0) # combining the viewership data for all three shows
show_viewership=show_viewership.drop(columns=['index'])
print(show_viewership)
```
