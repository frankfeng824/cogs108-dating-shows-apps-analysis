# Research Question

How is U.S. reality dating show popularity, measured by average per-episode ratings and per-season viewership, associated with changes in U.S. dating app activity (Tinder, Bumble, and Hinge), measured by annual revenue and annual user counts, from 2019–2026 using all available data?

## Background and Prior Work

The things that happen in dating shows are not like actual dating. As a form of reality TV that wishes to attrach a large, engaged audience, it's inevitable that producers and writers of dating shows artificialy infuse as much drama and tension as possible into each episode and season. Therefore, it shouldn't be a surprise that frequent viewers of dating shows tend to develop quite a warped perception of what romance should look like in everyday life. As early as 2007, a study already found that greater viewing of dating shows strongly predicts the tendencies to believe that "men are sex-driven," "dating is a game," and "women are sex objects."<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Unfortunately, such effects of dating shows have not changed much. A study looking at a much more recent dating show, "Singles Inferno," also found that the show perpetuates gender sterestypes and unrealistic beauty standards.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

Of course, dating apps, such as Tinder, Hinge, and Bumble, have also played quite a large role in the modern dating scene. The Pew Research Center finds that 30% of Americans have used dating apps at least once, and this number is particular higher for younger and queer people.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) However, just like dating shows, dating apps have also proven to be quite problematic. The same Pew Research Center report finds that over half of young women (18-34) report negative interactions while using dating apps, such as being sent unsolicited sexual content. Other studies find that dating app usage correlates with substance abuse and insecurity about physical appearances.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

Our project is motivated by the connections we see between dating shows and dating apps. They seem to both be related to the sexualization of women, superficial perceptions of romantic connection, and the perpetuation of unrealistic beauty standards. Although we are unable to run an experiment and establish any causal mechanisms between the two, we sense that the similar sociocultural sentiments underlying both dating shows and dating apps should be reflected by a correlation between dating show viewership/ratings and dating app usage. Beyond the aforementioned studies that look at either dating shows or dating apps seperately, we find very little prior empirical research on this topic. However, others have also previously speculated that both dating shows and dating apps (particularly Love Island and Tinder, respectively) collectively contribute to how intimacy has become more "performative" and "transactional" in the modern day.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) 

Given our research framing of sociocultural sentiment, we ideally would not want three dating shows that are overly similar to each other. We ended up choosing three shows that are relatively diverse in their premises and values. The first is Love Island USA, where multiple singles live together in a villa, couple up, get "dumped" by public vote, and reshuffle constantly, appears to magnify how dating is messy and full of drama. The second, in contrast, is The Bachelor, which magnifies how dating is competetive, since its premise is that one person dates and choose among dozens of other contestants simulatenously. Finally, the third show, Love is Blind, is also special, since participants aren't able to see one another and must rely on personality alone to assess each other, removing physical appearance and superficiality from the equation. Therefore, if we were to find correlations between certain shows and dating app usage, but not the same correlations for other dating shows, then there is an argument to be made that the values underlying those shows, specifically, are more tied to dating app usage. Of course, there were also practical considerations influencing our show selection. We were limited in finding shows with relatively accessible data.

# Hypothesis


We expect a positive relationship between the popularity of U.S. reality dating shows and dating app activity in the U.S. from 2019 to 2026. Specifically, when a season has higher average viewership and ratings, we predict that user counts and annual revenue for apps like Tinder, Bumble, and Hinge will also increase in the same year. 

We make this prediction because dating shows and dating apps seem to have similar underlying sociocultural sentiments, as mentioned in the Background section. Additionally, dating shows may make dating more visible and normalized to viewers, making them wish to personally engage in dating behavior themselves. 

However, we do not expect this relationship to be perfectly and consistently apparent across all years, since there are other existing factors like economic conditions or app-specific changes, which may also affect user growth and revenue.

#### 2, Relationship Between Viewership and Ratings

Next, we test for any potential relationship between a show's viewership data and their ratings for a given season. Intuitively, we would expect a better show (i.e., with higher ratings) to attract more viewers, and we want to see if the data confirms this intuition. Perhaps more importantly, we need to make sure that there won't be any colinearity between our predictor variabes in our eventual model. That is, if we find a significant relationship between a show's annual viewership and ratings, we cannot use both variables in a linear model that predicts rating app usage in that same year

First, some data wrangling is necessary. We need to combine the viewership and ratings datasets, such that each observation (row) is meant for a single season of a show, and that season's viewership and rating should also be on that row. 

```python
# combine two datasets into one

ratings = ratings.rename(columns={'show name': 'show_name', 'season': 'season_num'})

season_ratings = ratings.groupby(['show_name', 'season_num'])['rating'].mean().reset_index()
season_ratings = season_ratings.rename(columns={'rating': 'average_rating'})

combined = pd.merge(viewership, season_ratings, on=['show_name', 'season_num'])
combined = combined.set_index('season_num')
combined = combined.drop(columns='above_2.5mil')

combined
```

Now that all relevant information has been organized into one single dataset, we now make a scatterplot with ratings on the x-axis and viewership on the y-axis. We include OLS best-fit lines with 95% confidence intervals to more directly show the relationship. 

```python
# for all seasons, regardless of which show they belong to

sns.regplot(
    data=combined,
    x='average_rating',
    y='num_viewers (mil)',
    scatter=False,
    ci=95,
    color='black',
    line_kws={'label': 'All shows'}
)
```

```python
# plot a line for shows seperately

sns.lmplot(
    data = combined,
    x = 'average_rating',
    y = 'num_viewers (mil)',
    hue = 'show_name',
    height = 6,
    aspect = 1.3,
    ci = 95
)

plt.title('Average Rating vs. Number of Viewers')
plt.xlabel('Average Rating')
plt.ylabel('Viewers (millions)')

plt.show()
```

This is quite unexpected. For all three shows, as well as when they are all combined, we find that a season's ratings and viewership are actually negatively correlated; a "better" show ends up attracting fewer viewers. One possible explanation for this finding might relate to systematic bias in who rates a show. Perhaps only a select subset of relatively dedicated viewers of a show watch the less popular seasons; since they are dedicated, they may have more favorable opinions towards the show overall, which leads to them leaving higher ratings, even for those less popular seasons. Less dedicated viewers are not present to drag the average rating down. This is somewhat akin to certain artists' most critically acclaimed songs having relatively few streams, while less acclaimed songs have more streams.

Finally, we calculate the Pearson's R coefficient for each correlation, as well as the p-value. However, note that the p-value, no matter its value, wouldn't be that persuasive anyways, given our limited sample size. 

```python
from scipy.stats import pearsonr

results = []

for show, show_data in combined.groupby('show_name'):
    r, p_value = pearsonr(
        show_data['average_rating'],
        show_data['num_viewers (mil)']
    )
    
    results.append({
        'show_name': show,
        'correlation': r,
        'p_value': p_value
    })

print(pd.DataFrame(results), "\n")

r, p_value = pearsonr(
    combined['average_rating'],
    combined['num_viewers (mil)']
)

print('All shows combined')
print('Correlation (r):', round(r, 3))
print('P-value:', round(p_value, 3))
```

It appears that none of these relationships are statistically significant, if we were to use the conventional 0.05 threshold. However, again, any p-value wouldn't be meaningful anyways in this situation. What's important is that the correlation coefficent indicates that the relationship between a show's rating and viewership within a season is quite strong. When we build our model to predict dating app usage/revenue, we need to make sure to not include both the same show's viewership data and its ratings data; we also need to be cautious when using viewership data from one show and ratings data from another show.
